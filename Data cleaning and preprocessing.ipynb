# Hotel Performance Analysis

```python
ig = px.scatter(df,x="Month",y="Room_Revenue",color="Month",title="Monthly Room Revenue")
ig.show()
```

### It help identity seasonal trends and months with higher or lower revenue.

# Room Occupancy analysis

```python
average = df["Available_Rooms"].mean()
fig = px.bar(df,x="Reserved_Rooms",y="Available_Rooms",color="Reserved_Rooms",title="Room Occupancy Analysis")
fig.show()
```

### It helps analyze room utilization and understand occupancy levels in the hotel

# Cost Analysis

```python
fig = px.pie(names=["Total_Costs","Fixes_Costs"],values=[df["Total_Costs"].sum(),df["Fixed_Costs"].sum()],title="Cost Analysis")
labels = ["Total_Costs", "Fixed_Costs"]
fig.show()
```

### It hepls understand how the overall hotel expenses are distributed among major coast categories
