# Hotel Performance Analysis 

```python
fig = px.scatter(df,x="Occupancy_Rate",y="Total_Revenue",color="Total_Revenue",size="Total_Revenue",title="Hotel Performance Analysis") 
fig.update_layout(xaxis_title="Occupancy Rate (%)",yaxis_title="Total_Revenue")
fig.show()
```

## Hotel Performance Analysis
- Evaluate overall hotel performance.
- Analyze key performance indicators.
- Compare revenue and occupancy trends.
- Identify strengths and improvement areas.
- Support better business decisions.

# Monthly Revenue Trend

```python
fig = px.line(df,x="Month",y="Total_Revenue",markers=True,title="Monthly Revenue trend")
fig.show()
```

# Monthly Room Revenue

```python
fig = px.bar(df,x="Month",y="Room_Revenue",text="Room_Revenue",color="Room_Revenue",title="Monthly Room Revenue",height=600)
fig.update_layout(title_x=0.5,xaxis_title="Month",yaxis_title="Room Revenue")
fig.show()
```

## It help identity seasonal trends and months with higher or lower revenue.
- Analyze monthly room revenue.
- Compare revenue across different months.
- Identify highest and lowest revenue months.
- Detect monthly revenue trends.
- Support revenue planning and business growth.  

# Room Occupancy analysis

```python
room_data={"Status":["Reserved Rooms","Available Rooms"],"Rooms":[df["Reserved_Rooms"].sum(),df["Available_Rooms"].sum()]}
fig = px.pie(room_data,names="Status",values="Rooms",title="Room Occupancy analysis")
fig.update_layout(title_x=0.5)
fig.show()
```

## It helps analyze room utilization and understand occupancy levels in the hotel
- Analyze room occupancy levels.
- Compare reserved and available rooms.
- Identify peak occupancy periods.
- Detect room utilization trends.
- Support efficient room management.
