# Hotel Performance Analysis 

```python
fig= px.scatter(df,x="Room_Revenue",y="Total_Revenue",color="Season",size="Occupancy_Rate",hover_data="Date",
                title="Hotel Performance Analysis") 
fig.show()
```

###  Hotel Performance Analysis
#### 1.Compares Room Revenue with Total Revenue
#### 2. colored by Season and sized by Occupancy Rate to analyze hotel performance.

# Monthly Revenue Trend

```python
monthly = df.groupby("Month" , as_index=False)["Total_Revenue"].sum()
fig = px.area(monthly,x="Month",y="Total_Revenue",title="Monthly Revenue trend")
fig.update_traces(line_shape="spline",mode="lines+markers")
fig.update_layout(template="plotly_white")
fig.show()
```

### Monthly Revenue Trend
#### Displays the total revenue generated in each month and highlights seasonal revenue trends. 

#  Revenue by season 

```python
fig = px.bar(df,x="Season",y="Total_Revenue",color="Season",title="Hotel Cost Analysis by Season")
fig.update_layout(title_x=0.5)
fig.show()
```

## Season-wise Analysis
#### Compares total revenue across different seasons to identify the highest and lowest performing seasons.

# Cancellations Analysis trend

```python

c = df.groupby("Month")["Cancellations"].sum().reset_index()

px.line(c, x="Month", y="Cancellations",
        title="Cancellation Analysis Trend",
        markers=True).update_traces(line_color="red").show()
```

### Key Insight
#### 1.Month 1 has the highest number of cancellations, while Month 3 records the lowest cancellations.
#### 2.Cancellations decrease from Month 1 to Month 3 and then increase again by Month 7, showing a changing cancellation trend.

# Average Revenue:Weekday 

```python
fig = px.bar(df,x="Weekday",y="Total_Revenue",color="Total_Revenue",color_continuous_scale="viridis",title="Average Revenue Weekday")
fig.show()
```

### Key Insight
#### 1.Weekday 2 records the highest total revenue, while Weekdays 3, 4, and 7 generate comparatively lower revenue.
#### 2.Revenue remains fairly consistent across all weekdays, indicating stable hotel performance throughout the week.

# Monthly Guest type analysis

```python
plt.figure(figsize=(12,6))
sns.countplot(df,x="Month",hue="Guest_Type")
plt.title("Monthly Guest type analysis")
plt.xlabel("Month")
plt.ylabel("Number of Guests")
plt.show()
```

### Key Insight
#### 1.Leisure guests outnumber business guests in most months, indicating they make up the majority of hotel bookings.
#### 2.Month 1 has the highest number of leisure guests, while Month 2 shows nearly equal numbers of leisure and business guests.

# Monthly Booking trend

```python
monthly_bookings = df.groupby("Month")["Bookings"].sum().reset_index()
fig = px.bar(monthly_bookings,x="Month",y="Bookings",color="Bookings",color_continuous_scale="Pinkyl",title="Monthly Booking Trend")
fig.show()
```

### Key Insights
#### 1.Month 1 has the highest number of bookings, while Month 3 records the lowest bookings among the available months.
#### 2.Booking volume remains high in Months 1, 2, and 7, indicating stronger customer demand during these periods.

# Revenue by Channel(Hotel Substitute)

```python
booking_Channel = df.groupby("Booking_Channel")["Total_Revenue"].sum().reset_index()
fig = px.bar(booking_Channel, x="Booking_Channel",y="Total_Revenue",color="Booking_Channel",title="Revenue by channel(Hotel Substitute)")
fig.show()
```

### Key Insight
#### 1.Shows the total revenue generated through each booking channel, allowing comparison of channel performance.
#### 2.The Direct booking channel generates higher revenue than the OTA booking channel.

# Top Revenue by country(City Substitute)

```python
c = df.groupby("Guest_Country")["Total_Revenue"].sum().reset_index()
px.bar(c,x="Total_Revenue",y="Guest_Country",color="Total_Revenue",title="Top 10 Revenue by country(City Substitute)",
       orientation="h",color_continuous_scale="Blues").show()
```

### Key Insight
#### 1.Shows the total revenue generated by each guest country, helping compare country-wise revenue performance.
#### 2.Longer bars indicate higher revenue, while shorter bars indicate lower revenue contribution.

# Revenue By Market segment

```python

s = df.groupby("Market_Segment")["Total_Revenue"].sum().reset_index()

px.pie(s, names="Market_Segment", values="Total_Revenue",
       hole=0.4, title="Revenue by Market Segment").show()
```

### Key Insight
#### 1.Shows the contribution of each market segment to the total revenue, making it easy to compare revenue shares.
#### 2.Larger slices represent higher revenue contribution, while smaller slices indicate lower revenue contribution.

# Overall Costs distribution

```python
profit= df.groupby("Date")["Profit"].sum().reset_index()
fig = px.bar(profit,x="Date",y="Profit",color="Profit",title="Overall Costs Distribution")
fig.show()
```

### KEY Insight
#### 1. Shows the total profit earned on each date, helping identify daily profit performance.
#### 2.Higher bars and yellow colors indicate higher profit, while shorter blue/purple bars indicate lower profit.
